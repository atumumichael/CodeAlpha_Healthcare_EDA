```python
import matplotlib.pyplot as plt

condition_counts = df['Medical Condition'].value_counts()

plt.figure(figsize=(10, 6))

bars = plt.bar(
    condition_counts.index,
    condition_counts.values
)

plt.title(
    'Patient Distribution by Medical Condition',
    fontsize=14,
    fontweight='bold'
)

plt.xlabel('Medical Condition')
plt.ylabel('Number of Patients')

plt.bar_label(bars, padding=3)

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

age_counts = df['Age Group'].value_counts().sort_index()

plt.figure(figsize=(10, 6))

bars = plt.bar(age_counts.index.astype(str), age_counts.values)

plt.title(
    'Patient Distribution by Age Group',
    fontsize=14,
    fontweight='bold'
)

plt.xlabel('Age Group')
plt.ylabel('Number of Patients')

plt.bar_label(bars, padding=3)

plt.tight_layout()
plt.show()
```

```python
admission_counts = df['Admission Type'].value_counts()

plt.figure(figsize=(9, 6))

bars = plt.bar(
    admission_counts.index,
    admission_counts.values
)

plt.title(
    'Patient Distribution by Admission Type',
    fontsize=14,
    fontweight='bold'
)

plt.xlabel('Admission Type')
plt.ylabel('Number of Patients')

plt.bar_label(bars, padding=3)

plt.tight_layout()
plt.show()
```

```python
billing_by_condition = (
    df.groupby('Medical Condition')['Billing Amount']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

bars = plt.bar(
    billing_by_condition.index,
    billing_by_condition.values
)

plt.title(
    'Average Billing by Medical Condition',
    fontsize=14,
    fontweight='bold'
)

plt.xlabel('Medical Condition')
plt.ylabel('Average Billing Amount')

plt.bar_label(bars, fmt='%.0f', padding=3)

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

# Healthcare Patient Data Analysis Using Python

## 1. Project Overview

This project explores a healthcare dataset using Python to identify patterns, distributions and relationships in patient demographics, medical conditions, hospital admissions, billing amounts and hospital stays.

The project was completed as part of the CodeAlpha Data Analytics Internship.

## 2. Project Objectives

- Examine patient demographics and age distribution.
- Identify the most frequently recorded medical conditions.
- Analyse hospital admission types.
- Investigate billing patterns.
- Calculate hospital length of stay.
- Explore relationships between patient age, admission type and billing.
- Present findings using data visualizations.

## 3. Tools and Technologies

- Python
- Google Colab
- Pandas
- Matplotlib
- Kaggle Healthcare Dataset

## 4. Dataset Overview

The original dataset contained 55,500 records and 15 columns.

After removing 534 duplicate records, the cleaned dataset contained 54,966 records.

The dataset includes patient age, gender, blood type, medical condition, admission date, discharge date, hospital, insurance provider, billing amount, medication and test results.

## 5. Data Cleaning

- Checked for missing values.
- Identified and removed 534 duplicate records.
- Investigated 106 records with negative billing amounts.
- Converted admission and discharge dates into datetime format.
- Created age groups.
- Calculated hospital length of stay.

## 6. Analysis and Visualization

The analysis examined patient demographics, medical conditions, admission types, gender distribution, billing amounts, medication, test results and hospital length of stay.

Eight visualizations were created using Matplotlib.

## 7. Conclusion

The analysis revealed several patterns in the dataset, including a larger representation of patients aged 66–89, an almost balanced gender distribution and relatively similar average hospital stays across admission types.

The results provide a descriptive overview of the dataset and demonstrate how Python can be used to explore healthcare-related data.

## 8. Key Findings

### 8.1 Patient Demographics
- The average patient age was approximately 51.54 years.
- Patients aged 66–89 formed the largest age group, with 16,096 records.
- Male and female patients were almost equally represented.

### 8.2 Medical Conditions
- Arthritis was the most frequently recorded condition, with 9,218 patients.
- Asthma had the lowest count, with 9,095 patients.
- All six medical conditions had relatively similar record counts.

### 8.3 Hospital Admissions
- Elective admissions were the most frequent, with 18,473 records.
- Emergency admissions were the least frequent, with 18,102 records.
- Average hospital stay was approximately 15.50 days.
- Emergency admissions had the highest average stay at 15.58 days.

### 8.4 Billing Analysis
- Average billing across all records was approximately 25,544.31.
- Obesity had the highest average billing by medical condition.
- Patients aged 13–19 had the highest average billing by age group.
- A total of 106 records contained negative billing amounts.

### 8.5 Medication and Test Results
- Lipitor was the most frequently recorded medication.
- Abnormal results were slightly more frequent than normal and inconclusive results.
- Both medication and test-result categories had relatively balanced distributions.

## 9. Recommendations

1. Healthcare administrators should investigate billing records with negative values to determine whether they represent refunds, adjustments or data-entry issues.
2. Hospitals should monitor patient length of stay to identify opportunities for improving operational efficiency.
3. Billing patterns should be examined alongside other variables before drawing conclusions about healthcare costs.
4. Future analysis could investigate relationships between medical conditions, medications, test results and length of stay.

## 10. Limitations

- The dataset is a public Kaggle dataset and may not represent actual hospital populations.
- The analysis identifies associations and distributions, not causal relationships.
- Negative billing values require further investigation.
- The dataset does not provide enough context to explain why individual patients received particular treatments or incurred particular bills.

## 11. Final Conclusion

This project demonstrated the use of Python for healthcare data exploration, cleaning, statistical analysis and visualization.

Using Pandas and Matplotlib, 54,966 records were analysed after duplicate removal. The analysis identified patterns in patient demographics, medical conditions, admissions, billing and hospital length of stay.

The project demonstrates how data analytics can support the exploration and interpretation of healthcare datasets.
